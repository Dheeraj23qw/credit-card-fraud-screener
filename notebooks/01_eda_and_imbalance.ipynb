# Project 4: Credit Card Fraud Detection — EDA & Extreme Imbalance Analysis

## 1. Problem Overview
In financial fraud detection, genuine transactions outnumber fraud by over 50 to 1 (~1-2% or less). Standard accuracy metric is misleading because a dummy classifier predicting 100% legitimate transactions achieves 98-99% accuracy while catching 0% of fraud attacks.

**Goals:**
1. Explore distributions of transaction amount, time-of-day, distance, and velocity.
2. Analyze class imbalance and the inadequacy of accuracy as a metric.
3. Engineer features tailored to anomaly detection.

```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (12, 5)

sys.path.append(os.path.abspath("../src"))
from data_loader import load_raw_data, clean_data

df_raw = load_raw_data()
print(f"Total transactions: {len(df_raw)} | Fraud instances: {df_raw['Is_Fraud'].sum()} ({df_raw['Is_Fraud'].mean():.2%})")
df_raw.head()
```

## 2. Fraud Behavioral Patterns

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Hour of Day vs Fraud
sns.kdeplot(data=df_raw[df_raw["Is_Fraud"] == 1]["Hour_of_Day"], label="Fraudulent (1)", color="red", fill=True, ax=axes[0])
sns.kdeplot(data=df_raw[df_raw["Is_Fraud"] == 0]["Hour_of_Day"], label="Legitimate (0)", color="green", fill=True, ax=axes[0])
axes[0].set_title("Transaction Hour Density by Fraud Status", fontsize=13)
axes[0].legend()

# Distance from Home vs Fraud
sns.boxplot(data=df_raw, x="Is_Fraud", y="Distance_from_Home_KM", ax=axes[1], palette=["#10B981", "#EF4444"])
axes[1].set_title("Distance from Home (KM) by Fraud Status", fontsize=13)
axes[1].set_yscale("log")

plt.tight_layout()
plt.show()
```
